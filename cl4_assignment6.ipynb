```python
# Read dataset file
with open("word-freq.txt", "r") as file:
    data = file.readlines()

print("===== INPUT DATA =====")
for line in data:
    print(line.strip())
    
# MAP PHASE
# Create (word, 1) key-value pairs

mapped_data = []

for line in data:
    
    # Convert text to lowercase
    line = line.lower()
    
    # Remove punctuation
    line = line.replace(",", "")
    line = line.replace(".", "")
    
    # Split into words
    words = line.split()
    
    # Generate key-value pairs
    for word in words:
        mapped_data.append((word, 1))

print("\n===== MAP OUTPUT =====")
for pair in mapped_data:
    print(pair)
    
# REDUCE PHASE
# Directly count word frequencies

word_count = {}

for word, count in mapped_data:
    
    if word in word_count:
        word_count[word] += count
    else:
        word_count[word] = count

print("\n===== FINAL WORD FREQUENCY =====")
for word, frequency in word_count.items():
    print(word, ":", frequency)
```
