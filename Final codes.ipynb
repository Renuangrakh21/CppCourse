### CL3 Assignment 9- DC
Design and develop a distributed application to find the coolest/hottest year from the available weather data. Use weather data from the Internet and process it using MapReduce.


```python
import csv
import matplotlib.pyplot as plt
from collections import defaultdict

# ---------- MAP + SHUFFLE + REDUCE ----------
grouped = defaultdict(list)
with open("weather.csv", 'r') as f:
    for row in csv.DictReader(f):
        try:
            grouped[row['Date.Year']].append(float(row['Data.Temperature.Avg Temp']))
        except (KeyError, ValueError):
            pass

year_avg = {year: sum(t)/len(t) for year, t in grouped.items()}

# ---------- RESULT ----------
print("Hottest Year:", max(year_avg, key=year_avg.get), "Avg Temp:", max(year_avg.values()))
print("Coolest Year:", min(year_avg, key=year_avg.get), "Avg Temp:", min(year_avg.values()))

# ---------- BAR CHART ----------

plt.figure()
plt.bar(years, avg_temps,color=['tomato', 'steelblue'])
plt.xlabel("Year")
plt.ylabel("Average Temperature")
plt.title("Year vs Average Temperature")
plt.xticks(rotation=45)
plt.show()
```

### CL4 Assignment 5- BDA
Perform the data clustering algorithm using any Clustering algorithm.

```python
from sklearn.datasets import load_iris
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt

# Load Iris dataset
iris = load_iris()
X = iris.data

# Apply K-Means
kmeans = KMeans(n_clusters=3)
kmeans.fit(X)

# Get centroids and labels
centroids = kmeans.cluster_centers_
labels = kmeans.labels_

# Plot clusters
plt.scatter(X[:, 0], X[:, 1], c=labels)
plt.scatter(centroids[:, 0], centroids[:, 1], marker='X', s=200)
plt.title("K-Means Clustering")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.show()
```
